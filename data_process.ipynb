```python
# Importamos las librerías necesarias
import os                   # Para crear carpetas y manejar rutas de archivos
import requests             # Para hacer peticiones a internet y descargar los archivos
import zipfile              # Para descomprimir archivos .zip
import io                   # Para manejar los archivos descargados en la memoria
import pandas as pd         # Para el manejo y análisis de datos (DataFrames)
import matplotlib.pyplot as plt # Para crear gráficos básicos
import seaborn as sns       # Para crear gráficos más estéticos y estadísticos

# Configuramos el estilo visual por defecto para las gráficas de Seaborn
sns.set_theme(style="whitegrid")
```

```python
# 1. Creamos la carpeta "Datos" si no existe
carpeta_destino = "Datos"
if not os.path.exists(carpeta_destino):
    os.makedirs(carpeta_destino)
    print(f"Carpeta '{carpeta_destino}' creada.")

# 2. Definimos las URLs de los archivos a descargar
# Nota: Debes reemplazar estos enlaces de ejemplo por las URLs directas de los .zip
# que se encuentran en el Portal SAT para los meses de 2024 y 2025.
urls_archivos = ["https://portal.sat.gob.gt/portal/descarga/5030/importacion-de-vehiculos/75808/importacion_de_vehiculos_2024_enero",
    "https://portal.sat.gob.gt/portal/descarga/5030/importacion-de-vehiculos/76897/importacion_de_vehiculos_2024_febrero",
    "https://portal.sat.gob.gt/portal/descarga/5030/importacion-de-vehiculos/77313/importacion_de_vehiculos_2024_marzo",
    "https://portal.sat.gob.gt/portal/descarga/5030/importacion-de-vehiculos/78234/importacion_de_vehiculos_2024_abril",
    "https://portal.sat.gob.gt/portal/descarga/5030/importacion-de-vehiculos/79093/importacion_de_vehiculos_2024_mayo",
    "https://portal.sat.gob.gt/portal/descarga/5030/importacion-de-vehiculos/79819/importacion_de_vehiculos_2024_junio",
    "https://portal.sat.gob.gt/portal/descarga/5030/importacion-de-vehiculos/80559/importacion_de_vehiculos_2024_julio",
    "https://portal.sat.gob.gt/portal/descarga/5030/importacion-de-vehiculos/81358/importacion_de_vehiculos_2024_agosto",
    "https://portal.sat.gob.gt/portal/descarga/5030/importacion-de-vehiculos/81634/importacion_de_vehiculos_2024_septiembre",
    "https://portal.sat.gob.gt/portal/descarga/5030/importacion-de-vehiculos/82449/importacion_de_vehiculos_2024_octubre",
    "https://portal.sat.gob.gt/portal/descarga/5030/importacion-de-vehiculos/82785/importacion_de_vehiculos_2024_noviembre",
    "https://portal.sat.gob.gt/portal/descarga/5030/importacion-de-vehiculos/83292/importacion_de_vehiculos_2024_diciembre",
    "https://portal.sat.gob.gt/portal/descarga/5030/importacion-de-vehiculos/84293/importacion_de_vehiculos_2025_enero",
    "https://portal.sat.gob.gt/portal/descarga/5030/importacion-de-vehiculos/85486/importacion_de_vehiculos_2025_febrero",
    "https://portal.sat.gob.gt/portal/descarga/5030/importacion-de-vehiculos/85782/importacion_de_vehiculos_2025_marzo",
    "https://portal.sat.gob.gt/portal/descarga/5030/importacion-de-vehiculos/86197/importacion_de_vehiculos_2025_abril",
    "https://portal.sat.gob.gt/portal/descarga/5030/importacion-de-vehiculos/87242/importacion_de_vehiculos_2025_mayo"
]

# Cabecera para simular que somos un navegador web común y evitar bloqueos de la página
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'}

# 3. Iteramos sobre cada enlace para descargar y descomprimir
for i, url in enumerate(urls_archivos, start=1):
    try:
        print(f"Descargando archivo {i} desde {url}...")
        respuesta = requests.get(url, headers=headers)
        
        # Si la respuesta es 200 (OK), procedemos a descomprimir
        if respuesta.status_code == 200:
            # Leemos el archivo en memoria y lo descomprimimos directamente en 'Datos'
            archivo_zip = zipfile.ZipFile(io.BytesIO(respuesta.content))
            archivo_zip.extractall(carpeta_destino)
            print(f"Archivo {i} descomprimido con éxito.")
        else:
            print(f"Error al descargar. Código de estado: {respuesta.status_code}")
    except Exception as e:
        print(f"Ocurrió un error con el enlace {url}: {e}")
        
print("Proceso de descarga y extracción finalizado.")
```
